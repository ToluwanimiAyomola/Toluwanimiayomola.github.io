```python

import pandas as pd
import matplotlib.pyplot as plt

print("Nigeria Fuel Price Analysis")
```

```python

data = {
    "State": ["Lagos", "Abuja", "Kano", "Rivers", "Oyo", "Enugu", "Kaduna", "Ogun"],
    "January": [950, 970, 980, 960, 940, 965, 975, 945],
    "February": [980, 1000, 1010, 990, 970, 995, 1005, 975],
    "March": [1020, 1040, 1050, 1030, 1010, 1035, 1045, 1015],
    "April": [1050, 1070, 1080, 1060, 1040, 1065, 1075, 1045]
}

df = pd.DataFrame(data)

df
```

```python

df["Average Price"] = df[["January", "February", "March", "April"]].mean(axis=1)

df[["State", "Average Price"]]
```

```python

highest = df.loc[df["Average Price"].idxmax()]
lowest = df.loc[df["Average Price"].idxmin()]

print("Highest average price:")
print(highest[["State", "Average Price"]])

print("\nLowest average price:")
print(lowest[["State", "Average Price"]])
```

```python

monthly_average = df[["January", "February", "March", "April"]].mean()

print("Average fuel price by month:")
print(monthly_average)
```

```python

monthly_average.plot(kind="bar")

plt.title("Average Fuel Price by Month")
plt.xlabel("Month")
plt.ylabel("Average Price (₦)")
plt.xticks(rotation=0)
plt.show()
```

```python

plt.figure(figsize=(10, 5))

plt.bar(df["State"], df["Average Price"])

plt.title("Average Fuel Price by State")
plt.xlabel("State")
plt.ylabel("Average Price (₦)")
plt.xticks(rotation=45)

plt.show()
```

```python

df["Price Increase"] = df["April"] - df["January"]

df[["State", "Price Increase"]]
```

```python

biggest_increase = df.loc[df["Price Increase"].idxmax()]

print("State with the biggest price increase:")
print(biggest_increase[["State", "Price Increase"]])
```

```python
print("PROJECT SUMMARY")
print("----------------")
print("Number of states:", len(df))
print("Highest average price:",
      df.loc[df["Average Price"].idxmax(), "State"])
print("Lowest average price:",
      df.loc[df["Average Price"].idxmin(), "State"])
print("Biggest price increase:",
      df.loc[df["Price Increase"].idxmax(), "State"])
```
